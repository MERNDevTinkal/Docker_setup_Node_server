````markdown
# Configuration Management – Assignment

---

## Q1. Research and compare different configuration management tools (Ansible, Puppet, Chef) and write a report on their use cases and advantages.

---

### **Answer:**

### 1. Ansible
- **Type:** Agentless, uses SSH.  
- **Language:** YAML (Ansible Playbooks).  
- **Advantages:**  
  - Easy to learn and use.  
  - No agents required on target systems.  
  - Large community support.  
- **Use Cases:**  
  - Automating app deployment.  
  - Cloud provisioning (AWS, Azure, GCP).  
  - Orchestrating multi-tier environments.

---

### 2. Puppet
- **Type:** Agent-based (uses Puppet Master and Agents).  
- **Language:** Domain Specific Language (DSL, Puppet scripts).  
- **Advantages:**  
  - Strong for **managing large-scale infrastructure**.  
  - Good reporting and auditing.  
  - Mature tool with enterprise support.  
- **Use Cases:**  
  - Managing thousands of servers in data centers.  
  - Ensuring compliance and enforcing configurations.  

---

### 3. Chef
- **Type:** Agent-based (Chef Server + Nodes).  
- **Language:** Ruby DSL (Recipes & Cookbooks).  
- **Advantages:**  
  - Highly flexible and customizable.  
  - Strong integration with cloud providers.  
  - Suitable for complex infrastructure.  
- **Use Cases:**  
  - Automating server builds.  
  - Cloud deployments (AWS OpsWorks).  
  - Custom infrastructure automation at scale.

---

### 4. Comparison Table

| Feature                  | Ansible                | Puppet                     | Chef                       |
|---------------------------|------------------------|----------------------------|----------------------------|
| Architecture              | Agentless (SSH)        | Master-Agent               | Master-Agent               |
| Language                  | YAML                   | DSL (Puppet)               | Ruby DSL                   |
| Learning Curve            | Easy                   | Medium                     | High                       |
| Best For                  | Small → Medium setups  | Large-scale infra mgmt     | Complex infra & custom use |
| Community Support         | Very Strong            | Strong                     | Moderate                   |
| Cloud Integration         | Good                   | Good                       | Excellent                  |

---

###  Conclusion:
- **Ansible** → Best for beginners, quick automation, and cloud provisioning.  
- **Puppet** → Best for enterprise-level, large infrastructure, compliance-heavy use.  
- **Chef** → Best when you need complex, customized automation with deep cloud integration.  

---

## Q2. Extend the CI pipeline to include continuous delivery using Jenkins or GitLab CI. Implement a blue-green deployment strategy.

---

### **Answer:**

### 1. Continuous Delivery with Jenkins
- Modified Jenkins pipeline to:  
  1. Build application.  
  2. Run unit & integration tests.  
  3. Deploy to **staging environment**.  
  4. If staging is successful → automatically deploy to production.

**Sample Jenkinsfile**
```groovy
pipeline {
  agent any
  stages {
    stage('Build') {
      steps {
        sh 'mvn clean package'
      }
    }
    stage('Test') {
      steps {
        sh 'mvn test'
      }
    }
    stage('Deploy to Staging') {
      steps {
        sh './deploy.sh staging'
      }
    }
    stage('Blue-Green Deployment') {
      steps {
        sh './blue_green_deploy.sh production'
      }
    }
  }
}
````

---

### 2. Blue-Green Deployment Strategy

* **Concept:** Maintain **two environments** (Blue & Green).

  * Blue = Current live environment.
  * Green = New version deployed in parallel.
* **Steps:**

  1. Deploy new version to **Green**.
  2. Run tests on Green.
  3. Switch traffic from **Blue → Green**.
  4. Keep Blue as backup (rollback option).

**Sample Script (`blue_green_deploy.sh`):**

```bash
#!/bin/bash
APP=$1

if [ "$APP" == "production" ]; then
  echo "Deploying new version to GREEN environment..."
  docker-compose -f docker-compose.green.yml up -d

  echo "Running smoke tests..."
  curl -f http://green-env/health || exit 1

  echo "Switching traffic from BLUE to GREEN..."
  # Update load balancer to point to Green
  aws elbv2 modify-listener --listener-arn arn:xyz --default-actions TargetGroup=green

  echo "Deployment successful. BLUE is idle (rollback option)."
fi
```

### 3. GitLab CI Alternative

**`.gitlab-ci.yml`**

```yaml
stages:
  - build
  - test
  - deploy

build:
  script:
    - mvn clean package

test:
  script:
    - mvn test

deploy_blue_green:
  script:
    - ./blue_green_deploy.sh production
  only:
    - main
```

---

###  Conclusion:

* CI/CD pipeline extended to **Continuous Delivery**.
* Implemented **Blue-Green Deployment** to ensure zero downtime.
* Rollback is easy (switch traffic back to Blue if Green fails).
