# 🔹 What is Ansible?

 Ansible is an **Automation & Configuration Management tool**.

* Imagine you have **10 servers**. On each server you need to install **Docker + Nginx + Node.js**.
* If you do it manually: login one-by-one, type commands, repeat. Boring + Error-prone.
* With Ansible: you write steps in a **YAML file** (called *Playbook*), and Ansible executes them on **all servers at once** via SSH.

➡ In short:

* **Manual** → Type commands on each server
* **Ansible** → One script, runs on all

---

# 🔹 Why do we need Ansible?

1. **Agentless** → No software required on target servers (unlike Puppet, Chef). Only SSH.
2. **Idempotent** → Run the same script multiple times, server stays consistent.
3. **Easy** → Uses YAML (human-readable).
4. **Integration** → Works with Docker, Kubernetes, AWS, Jenkins, etc.
5. **DevOps use** → Infra automation, app deployment, config management.

---

# 🔹 How Ansible Works

* **Control Node** (your laptop/EC2 with Ansible installed)
* **Managed Nodes** (servers where you want changes)

Steps:

1. Install Ansible on Control Node
2. Define target servers in **Inventory file**
3. Write tasks in **Playbook (YAML)**
4. Run → Ansible connects via SSH → executes

---

# 🔹 Simple Example

 Suppose you want to install Nginx on 2 servers.

**Inventory file (hosts.ini):**

```ini
[web]
192.168.1.10
192.168.1.11
```

**Playbook (install-nginx.yml):**

```yaml
- name: Install Nginx on servers
  hosts: web
  become: yes
  tasks:
    - name: Install Nginx
      apt:
        name: nginx
        state: present
        update_cache: yes

    - name: Start Nginx
      service:
        name: nginx
        state: started
        enabled: yes
```

**Run command:**

```bash
ansible-playbook -i hosts.ini install-nginx.yml
```

 Result → Both servers will get Nginx installed & running automatically.
