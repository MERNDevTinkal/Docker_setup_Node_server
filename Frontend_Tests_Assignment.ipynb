* `-n` → non-GUI mode
* `-t` → test plan file
* `-l` → log results
* `-e -o` → generate HTML dashboard report

---

### 4. Sample Results (Summary Report)

| Metric            | Value       |
| ----------------- | ----------- |
| Concurrent Users  | 100         |
| Avg Response Time | 350 ms      |
| 90th Percentile   | 500 ms      |
| Errors (%)        | 2%          |
| Throughput        | 150 req/sec |

---

### 5. Analysis

* **Observation**:

  * Average response time (350ms) is acceptable.
  * 90th percentile (500ms) indicates some requests are slower.
  * Error rate (2%) suggests possible server overload at high concurrency.
* **Performance Bottlenecks Identified**:

  * Login API slows down with >80 concurrent users.
  * Possible database connection bottleneck.
  * Static assets (images/scripts) increase load time.

---

### 6. Bottleneck Fix Recommendations

* Implement **database connection pooling**.
* Use **CDN** for serving static assets.
* Enable **server-side caching** (Redis / Varnish).
* Optimize frontend (minify JS/CSS, lazy load images).
* Consider **horizontal scaling** using load balancers.

---

### 7. (Optional) Gatling Alternative

Using **Gatling (Scala/Java)** for code-driven load testing:

**Sample Gatling Script (`BasicSimulation.scala`)**

```scala
import io.gatling.core.Predef._
import io.gatling.http.Predef._
import scala.concurrent.duration._

class BasicSimulation extends Simulation {
  val httpConf = http.baseUrl("http://localhost:3000")

  val scn = scenario("Load Test Frontend")
    .exec(http("Homepage Request").get("/"))

  setUp(
    scn.inject(
      rampUsers(100) during (10 seconds)
    )
  ).protocols(httpConf)
}
```

Run:

```bash
mvn gatling:test
```
