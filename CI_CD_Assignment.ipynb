
## 1. Continuous Integration (CI)

**Q1. Set Up a CI Pipeline**

- **Objective**: Set up a basic CI pipeline that builds a project automatically whenever new code is pushed.
- **Answer**:  
  1. Created a simple Node.js app (`app.js`) with a basic "Hello World" API.  
  2. Pushed code to GitHub repository.  
  3. Created `.github/workflows/ci.yml` file:  

  ```yaml
  name: Node.js CI
  on: [push]
  jobs:
    build:
      runs-on: ubuntu-latest
      steps:
        - uses: actions/checkout@v2
        - name: Setup Node.js
          uses: actions/setup-node@v2
          with:
            node-version: '16'
        - run: npm install
        - run: npm test
````

4. Verified pipeline runs automatically on every push.

---

## 2. Jenkins

Q2. Automated Build with Jenkins**

* Objective: Install Jenkins and configure an automated build job.
* Answer:

  1. Installed Jenkins locally using Docker:

     ```bash
     docker run -p 8080:8080 -p 50000:50000 jenkins/jenkins:lts
     ```
  2. Configured Jenkins, installed Git + Node plugins.
  3. Created "Freestyle Job" → pulled code from GitHub.
  4. Added build steps (`npm install && npm test`).
  5. Configured webhook so job triggers automatically on push.

---

## 3. Testing Frameworks and Practices

**Q3. Write Unit Tests for an Application**

* Objective: Write unit tests and integrate them into CI pipeline.
* Answer:

  1. Built a calculator app in Node.js (`calculator.js`).
  2. Wrote Jest unit tests (`calculator.test.js`):

     ```javascript
     const calc = require('./calculator');
     test('adds 2 + 3 = 5', () => {
       expect(calc.add(2, 3)).toBe(5);
     });
     ```
  3. Integrated into GitHub Actions (CI runs tests automatically).

---

## 4. Integration Testing

Q4. Set Up Integration Testing for an API**

* Objective: Write integration tests for a REST API.
* Answer:

  1. Built REST API with Express.js + MongoDB.
  2. Created Postman collection with requests (`GET /users`, `POST /users`).
  3. Used Newma* to automate:

     ```bash
     newman run api-tests.postman_collection.json
     ```
  4. Configured Jenkins pipeline to run integration tests after deployment to staging.

---

## 5. Performance Testing

Q5. Automate Performance Testing in a CI/CD Pipeline**

* Objective: Automate performance testing.
* Answer:

  1. Installed Apache JMeter.
  2. Created JMeter test plan to load test `/api/users`.
  3. Automated in Jenkins pipeline:

     ```groovy
     stage('Performance Test') {
       steps {
         sh 'jmeter -n -t load_test.jmx -l results.jtl'
       }
     }
     ```
  4. Captured performance report graphs.

---

## 6. Security Testing

Q6. Implement Basic Security Testing in a CI Pipeline**

* Objective: Run security scan with OWASP ZAP or SonarQube.
* Answer:

  1. Installed SonarQube locally (`docker run -d -p 9000:9000 sonarqube`).
  2. Configured project in Jenkins with SonarQube plugin.
  3. On every push, pipeline runs `sonar-scanner` for static code analysis.
  4. Detected vulnerabilities and code smells.

---

## 7. Continuous Deployment (CD)

Q7. Automate Deployment to Staging Environment**

* Objective: Deploy app to staging automatically after CI tests pass.
* Answer:

  1. Containerized app with Docker:

     ```dockerfile
     FROM node:16
     WORKDIR /app
     COPY . .
     RUN npm install
     CMD ["npm", "start"]
     ```
  2. Pushed Docker image to Docker Hub.
  3. Configured Jenkins pipeline to deploy to AWS EC2 (staging) using Docker run.
```
